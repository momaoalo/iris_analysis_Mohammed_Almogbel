## Task 4: Explore and Analyze the Dataset

```python
# Q1: How many rows and columns are in the dataset?
df.shape
```

**Answer:** The dataset has 150 rows and 5 columns.

```python
# Q2: Are there any missing values? (count per column)
df.isnull().sum()
```

**Answer:** No, there are no missing values. Every column has 0 nulls.

**Answer:** The maximum petal length is 6.9 cm.

```python
# Q8: What is the minimum sepal width?
df["sepal_width"].min()
```

**Answer:** The minimum sepal width is 2.0 cm.

```python
# Q9: Mean of each numeric column for each species
df.groupby("species").mean()
```

```python
# Q10: Descriptive statistics
df.describe()
```

### Summary
1. **Data quality:** There are no missing values, and only 1 duplicate row, so the dataset is clean.
2. **Species:** There are 3 species (setosa, versicolor, virginica) with 50 flowers each, so the data is balanced.
3. **Key values:** The average sepal length is about 5.84 cm, the maximum petal length is 6.9 cm, and the minimum sepal width is 2.0 cm.
4. **Group comparison:** Virginica has the largest petal measurements and setosa the smallest, so petal size separates the species best.

## Task 7: Data Visualization

```python
# 7.0: Import matplotlib (pandas plots are Matplotlib objects)
import matplotlib.pyplot as plt
```

```python
# 7.1: Quick visual check - one line per numeric column
df.plot()
plt.show()
```

```python
# 7.2: Box plot of each measurement, grouped by species
df.boxplot(by="species", figsize=(10, 8))
plt.suptitle("")
plt.show()
```

```python
# 7.3: List the available plot methods
[method_name for method_name in dir(df.plot) if not method_name.startswith("_")]
```

```python
# 7.4: Scatter plot of petal size, one color per species
colors = {"setosa": "tab:blue", "versicolor": "tab:orange", "virginica": "tab:green"}
fig, ax = plt.subplots(figsize=(7, 5))
for name, group in df.groupby("species"):
    group.plot.scatter(x="petal_length", y="petal_width", ax=ax,
                       color=colors[name], label=name, alpha=0.6)
ax.set_xlabel("Petal length (cm)")
ax.set_ylabel("Petal width (cm)")
ax.set_title("Petal size by species")
plt.show()
```

```python
# 7.5: Histogram of each numeric column
df.hist(figsize=(10, 6), bins=15)
plt.show()
```

```python
# 7.6: Bar chart of mean measurements per species
ax = df.groupby("species").mean().plot.bar(figsize=(8, 5), rot=0)
ax.set_ylabel("Mean (cm)")
ax.set_title("Mean measurements by species")
plt.show()
```

```python
# 7.7: Save the bar chart to the output folder
ax.figure.savefig("output/species_means.png", dpi=100)
```

   ### Summary
   1. **Species separate by petal size:** setosa has petal length 1.0-1.9 cm, clearly below versicolor (3.0-5.1) and virginica (4.5-6.9), so it forms its own cluster in the scatter plot.
   2. **Overlap:** versicolor and virginica overlap slightly, but virginica is larger on average.
   3. **Box plots and bar chart:** both confirm that petal measurements differ the most between species.
   4. **Histogram:** petal length has two groups of values (small for setosa, larger for the others), which explains why its mean is lower than its median.
